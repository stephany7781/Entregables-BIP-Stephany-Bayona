```python
def clasificar_rating(rating):
    if rating >= 4:
        return 'Excelente'
    elif rating >= 3:
        return 'Bueno'
    elif rating >= 2:
        return 'Regular'
    else:
        return 'Bajo'

df['Rating category'] = df['Aggregate rating'].apply(clasificar_rating)
```

```python
df.head(3)
```

```python
#¿Existe realción entre el precio promedio y la calificación?

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Average Cost for two',
    y='Aggregate rating'
)

plt.title('Relación entre precio promedio y calificación')
plt.xlabel('Precio promedio para dos personas')
plt.ylabel('Calificación')
plt.show()
```

```python
df['Average Cost for two'].corr(df['Aggregate rating'])
```

Interpretación: 
El gráfico de dispersión no muestra una tendencia lineal clara entre el precio promedio y la calificación. Esto coincide con el coeficiente de correlación de 0.052, que indica una relación lineal muy débil entre ambas variables

```python
#¿Cuales son las ciudades con la mayor cantidad de restaurantes registrados?

restaurantes_ciudad = df.groupby('City')['Restaurant ID'].count()
restaurantes_ciudad = restaurantes_ciudad.sort_values(ascending=False)
restaurantes_ciudad.head(10)
```

```python
restaurantes_ciudad.head(10).plot(kind='barh')
plt.title('Ciudades con mayor cantidad de restaurantes registrados')
```

Interpretación: 
Se observa una concentración de restaurantes registrados en pocas ciudades, siendo New Delhi la más destacada con 5.473 restaurantes, superando por mucho a Gurgaon y Noida. Esto indica que estas ciudades tienen una representación mucho mayor dentro del conjunto de datos analizado, especialmente New Delhi.

```python
#¿Qué relación existe entre la disponibilidad de domicilio y la calificación de los restaurantes?

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Has Online delivery',
    y='Aggregate rating'
)

plt.title('Calificación según disponibilidad de entrega a domicilio')
plt.xlabel('Entrega a domicilio')
plt.ylabel('Calificación')
plt.show()
```

Interpretación:
Los restaurantes que ofrecen entrega a domicilio presentan una mayor concentración de sus calificaciones en valores positivos, principalmente entre 3 y 4. En cambio, los restaurantes que no ofrecen este servicio muestran una mayor dispersión en sus calificaciones. Esto sugiere que existe una diferencia en la distribución de las calificaciones según la disponibilidad de entrega a domicilio.

```python
#¿En qué rango de precio están ubicados la mayoría de restaurantes? 

restaurantes_precio = df['Price range'].value_counts().sort_index()

plt.figure(figsize=(8,5))
restaurantes_precio.plot(kind='bar')

plt.title('Cantidad de restaurantes por rango de precios')
plt.xlabel('Rango de precios')
plt.ylabel('Cantidad de restaurantes')
plt.xticks(rotation=0)
plt.show()
```

Interpretación:
La mayoría de los restaurantes del conjunto de datos se encuentran en el rango de precios más bajo, la cantidad de restaurantes disminuye progresivamente a medida que aumenta el rango de precios. Este resultado coincide con el gráfico de dispersión anterior, donde se observaba una alta concentración de restaurantes en los precios más bajos.

```python
#¿Existe una relación entre el rango de precios y el promedio de votos de los restaurantes?

votos_precio = df.groupby('Price range')['Votes'].mean()

plt.figure(figsize=(8,5))
votos_precio.plot(kind='bar')

plt.title('Promedio de votos según rango de precios')
plt.xlabel('Rango de precios')
plt.ylabel('Promedio de votos')
plt.xticks(rotation=0)
plt.show()
```

Interpretación:
Aunque el rango de precios 1 concentra la mayor cantidad de restaurantes, el promedio de votos más alto corresponde a los restaurantes del rango 3. Esto indica que tener una mayor cantidad de restaurantes en un rango de precios no necesariamente implica recibir una mayor cantidad de votos en promedio. En este conjunto de datos, los restaurantes de rango 3 presentan una mayor participación promedio de los usuarios.

```python
#¿Los restaurantes con mejores calificaciones reciben más votos?

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Aggregate rating',
    y='Votes'
)

plt.title('Relación entre calificación y cantidad de votos')
plt.xlabel('Calificación')
plt.ylabel('Cantidad de votos')
plt.show()
```

nterpretación: 
Se observa una tendencia positiva entre la calificación y la cantidad de votos. Los restaurantes con calificaciones más altas tienden a presentar una mayor cantidad de votos, aunque existe una considerable dispersión entre los valores. Esto indica que los restaurantes mejor calificados suelen tener mayor participación de los usuarios, pero una calificación alta no necesariamente implica recibir una gran cantidad de votos.
