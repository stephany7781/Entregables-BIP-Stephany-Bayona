```python
#crear base de datos 
estudiantes = [
    ("Ana Rodríguez", 20, 4.5),
    ("Juan Martínez", 22, 3.8),
    ("María González", 19, 4.2),
    ("Carlos Pérez", 21, 3.5),
    ("Laura Hernández", 20, 4.8),
    ("Pedro Ramírez", 23, 3.9),
    ("Sofía Torres", 18, 4.6),
    ("Andrés Castillo", 22, 3.2),
    ("Valentina Moreno", 19, 4.0),
    ("Luis Vargas", 21, 2.9),
    ("Camila Herrera", 20, 4.7),
    ("Daniel Mendoza", 24, 3.6)]

print(estudiantes)
```

```python
# mostrar la información 
for estudiante in estudiantes:
    nombre, edad, nota = estudiante
    print(f'{nombre} tiene {edad} años y obtuvo una nota de {nota}')
```

```python
#Calificación de estudiantes 
for estudiante in estudiantes:
    nombre, edad, nota = estudiante
    if nota >= 4.5: 
         print(f'{nombre} tuvo {nota} una calificación excelente!')
    elif  4.0 <= nota <=4.49:
         print(f'{nombre} tuvo {nota}una buena calificación')
    elif  3.0 <= nota <=3.99:
         print(f'{nombre} tuvo {nota} una calificación aceptable')
    else:
          print(f'{nombre} reprobó')

    
```

```python
#Promedio general 
sumas_notas = 0

for estudiante in estudiantes:
    nombre, edad, nota = estudiante
    sumas_notas += nota
    promedio = sumas_notas/len(estudiantes)

print('el promedio obtenido de todas las notas es',promedio)
```

```python
#Busqueda de estudiantes 
nombre_buscar = input('Ingresa el nombre del estudiante:' )
encontrado = False 
for estudiante in estudiantes:
    nombre, edad, nota = estudiante
    if nombre == nombre_buscar: 
        print('¡Estudiante encontrato!')
        encontrado = True
if encontrado == False: 
        print('no se encontró ningún estudiante con ese nombre')
```

```python
#Diccionario de ciudades 
ciudades_estudiantes = {
    "Ana Rodríguez": "Barranquilla",
    "Juan Martínez": "Bogotá",
    "María González": "Cartagena",
    "Carlos Pérez": "Medellín",
    "Laura Hernández": "Cali",
    "Pedro Ramírez": "Santa Marta",
    "Sofía Torres": "Barranquilla",
    "Andrés Castillo": "Bogotá",
    "Valentina Moreno": "Cartagena",
    "Luis Vargas": "Medellín",
    "Camila Herrera": "Cali",
    "Daniel Mendoza": "Santa Marta"
}

for nombre,ciudades in ciudades_estudiantes.items(): 
    print(f'{nombre} vieve en {ciudades}')
```

```python
#Cantidad de estudiantes por ciudad 

cantidad_ciudades = {}

for ciudad in ciudades_estudiantes.values():
    if ciudad in cantidad_ciudades:
        cantidad_ciudades[ciudad] += 1
    else:
        cantidad_ciudades[ciudad] = 1

print(cantidad_ciudades)
```

```python
#Ciclo while 
cantidad = 0
suma = 0

numero = int(input("Ingresa un número (0 para terminar): "))

while numero != 0:
    cantidad += 1
    suma += numero

    numero = int(input("Ingresa un número (0 para terminar): "))

promedio = suma / cantidad

print("Cantidad de números:", cantidad)
print("Suma total:", suma)
print("Promedio:", promedio)
```
